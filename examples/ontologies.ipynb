# Managing different ontologies

The package `mammos-entity` can in general be used with different ontologies. Together with the Python software the [magnetic materials domain ontology (MagMO)](https://emmo-repo.github.io/domain-magnetic-materials/) is packaged, but all EMMO-based ontologies are supported.

```python
import logging

import requests

import mammos_entity as me

logging.basicConfig(format="%(levelname)s:%(message)s", level=logging.DEBUG)
```

## Specifying different ontologies

It is possible to specify the ontology `versionIRI` strings we want to consider:

```python
onto = me.Ontology(
    iris=[
        "https://w3id.org/emmo/domain/magnetic-materials/0.0.5",
        "https://w3id.org/emmo/1.0.2",
        "https://w3id.org/emmo/domain/electrochemistry/0.37.2/",
    ],
)
```

```python
print(str(onto))
```

Without specifying IRIs, only MagMO is loaded.

```python
onto = me.Ontology()
```

```python
print(str(onto))
```

## Ontology initialization

Ontologies implement lazy loading and are initialized only when needed (for example, when defining an Entity).

```python
%%time
onto = me.Ontology()
```

```python
%%time
onto.Entity("Magnetization", 1.2)
```

After that, calling any entity is going to be fast:

```python
%%time
onto.Entity("Magnetization", 1.2)
```

```python
%%time
onto.Entity("CurieTemperature", 400)
```

Alternatively, we can initialize an ontology explicitly by specifying the argument `initialize=True` when creating the object or calling the `initialize()` method.

```python
%%time
me.Ontology(initialize=True)
```

```python
%%time
onto = me.Ontology()
onto.initialize()
```

## Using same ontology - different versions

If an entity appears in multiple loaded ontologies, we use the one from the ontologies that appear first:

```python
onto = me.Ontology(
    iris=[
        "https://w3id.org/emmo/1.0.2",
        "https://w3id.org/emmo/1.0.3",
    ],
)
```

```python
onto.iris  # most recent appears first
```

```python
mag = onto.Entity("ElementaryCharge", 4.0)
mag
```

```python
mag.ontology_iri
```

## Add ontology iris later

```python
onto = me.Ontology(
    iris=[
        "https://w3id.org/emmo/1.0.2",
    ],
)
```

```python
onto.iris
```

```python
onto.add_iri("https://w3id.org/emmo/1.0.3")
onto.iris
```

Adding an already included ontology will warn:

```python
onto.add_iri("https://w3id.org/emmo/1.0.3")
```

If the ontology was initialized, adding another IRI will load the corresponding ontology. If it was not initialized, it will not load.

## Local ontologies

It is possible to use local ontologies as well by specifying `file://` in the IRI (in the same way we specify `https://` for URLs).

```python
res = requests.get("https://w3id.org/emmo/1.0.3/inferred")
with open("emmo.ttl", "wb") as f:
    f.write(res.content)
res = requests.get("https://w3id.org/emmo/domain/electrochemistry/0.37.2/inferred")
with open("echem.ttl", "wb") as f:
    f.write(res.content)
```

```python
%%time
onto = me.Ontology(
    iris=[
        "file://emmo.ttl",
        "file://echem.ttl",
    ],
    initialize=True,
)
onto
```

```python
onto.iris
```

## `EntityCollection` objects and metadata

Now each `Entity` in an `EntityCollection` has its own ontology metadata.

```python
onto = me.Ontology(
    iris=[
        "https://w3id.org/emmo/domain/magnetic-materials/0.0.5",
        "https://w3id.org/emmo/1.0.3/inferred",
        "https://w3id.org/emmo/domain/electrochemistry/0.37.2/",
    ],
    initialize=True,
)
```

```python
ec = me.EntityCollection(
    Ms=onto.Entity("SpontaneousMagnetization", 1),
    Tc=onto.Entity("ThermodynamicTemperature", 2),
    DC=onto.Entity("DischargingCapacity", 3),
)
metadata = ec.metadata()
metadata
```

The reading and writing functions function as before. Reading routines now define single Ontology objects for each different ontology IRI. If two entities have the same `ontology_iri`, they share the Ontology object.

```python
me.EntityCollection.from_dataframe(ec.to_dataframe(), metadata).metadata()
```
