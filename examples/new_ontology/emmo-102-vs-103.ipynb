```python
from ontopy import World
```

```python
emmo102 = World().get_ontology("https://w3id.org/emmo/1.0.2").load()
```

```python
emmo103 = World().get_ontology("https://w3id.org/emmo/1.0.3").load()
```

```python
str(emmo102.Magnetization.get_preferred_label())
```

```python
emmo102.get_by_label("Magnetization").iri
```

```python
emmo102.Magnetization == emmo103.Magnetization
```

## In 102 not in 103

```python
for thing102 in emmo102.classes():
    label = str(thing102.get_preferred_label())
    if label in emmo103 and thing102.iri != emmo103.get_by_label(label).iri:
        print(label)
```

```python
emmo102.ElementaryCharge.iri
```

```python
emmo103.ElementaryCharge.iri
```

```python
emmo102.ElementaryCharge == emmo103.ElementaryCharge
```

## In 103 not in 102

```python
for thing103 in emmo103.classes():
    label = str(thing103.get_preferred_label())
    if label not in emmo102:
        print(label)
```
